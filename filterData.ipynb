```python
import numpy as np
import pandas as pd
import os
```

```python
Data_Dir = "/tmp/pandas_class/"
os.makedirs(Data_Dir,exist_ok=True)
```

```python
np.random.seed(1)
n = 200
price_list = {"Laptop": 55000, "Mouse": 700, "Chair": 4500, "Desk": 9000, "Notebook": 60}
category_list = {"Laptop": "Electronics", "Mouse": "Electronics", "Chair": "Furniture",
                 "Desk": "Furniture", "Notebook": "Stationery"}

products = np.random.choice(list(price_list), n)

sales = pd.DataFrame({
    "order_id": np.arange(1001, 1001 + n),
    "order_date": np.random.choice(pd.date_range("2026-01-01", "2026-06-30"), n),
    "customer": np.random.choice(["Amit", "Sara", "John", "Priya", "Ravi", "Neha"], n),
    "city": np.random.choice(["Pune", "Mumbai", "Delhi", "Chennai"], n),
    "product": products,
    "category": [category_list[p] for p in products],
    "qty": np.random.randint(1, 6, n),
    "price": [price_list[p] for p in products],
})
sales["order_date"] = sales["order_date"].dt.strftime("%Y-%m-%d")
sales.to_csv(f"{Data_Dir}/sales.csv", index=False)

print("Created:", f"{Data_Dir}/sales.csv")
print("Rows:", len(sales), " Columns:", len(sales.columns))
```

```python
sales["amount"] = sales["qty"] * sales["price"]
```

```python
# Show only the columns `order_id`, `customer`, `amount` for the first 5 rows.
sales[["order_id","customer","amount"]].head(5)
```

```python
# All orders from Delhi with quantity 3 or more.
Delhi =  sales[(sales["city"] == "Delhi") & (sales["qty"] > 3)]
print(Delhi)
```

```python
# All orders of `Laptop` or `Desk`, sorted by amount (highest first). Show the top 5.
sales = sales[sales["product"].isin(["Laptop","Desk"])].sort_values(["amount"],ascending=True)
sales.head(5)
```

```python
# Orders where the amount is between 10,000 and 50,000. How many are there?
sales[sales["amount"].between(10000,50000)]
```

```python
# The 3 biggest orders - print customer, product and amount.
sales.nlargest(3,"amount")[["customer","product","amount"]]
```

```python
# Orders placed in February 2026
sales[(sales["order_date"] >= "2026-02=01") | (sales["order_date"] <= "2026-03-01")]
```

```python
# Same as question 2, but written with `query()`
# All orders from Delhi with quantity 3 or more.
sales.query("city == 'Delhi' and qty >= 3")
```

```python
# Make a Pune-only table, add a column `amount_after_10pc_off`, and print the first 3 rows
pune = sales[sales["city"] == "Pune"].copy()
pune["amount_after_10pc_off"] = pune["amount"] * 0.9
pune.head(3)
   
```
