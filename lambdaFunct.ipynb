### How to use lambda funvtion in different way

```python
#lambda is a anonymous small function defined using lambda keyword.
# it can take any number of arguments 
# but contains only one expression. 
#We use it when we need short function especially with functons
# like map(),filter(),sort(), and apply()
```

```python
# normal function
def square(x):
    return x**2
```

```python
square(5)
```

```python
#same function with lambda
sq = lambda x: x**2
sq(5)
```

```python
#lambda funtion with multiple arg
sum = lambda a,b : a+b
print("The sum of a and b is ",sum(3,7))
```

```python
even = lambda x:x%2==0
print(even(9))
even(6)
```

```python
#lambda with f else
greater = lambda a,b : f"{a} is greater than {b}" if a > b else f"{b} is greter than {a}"
```

```python
greater(0,9)
```

```python
#find the even no. in a list
num  = [2,5,8,54,68,24]
even = [map(lambda x:x%2==0 , num)]
print(even)
```

```python
num  = [2,5,8,54,68,24]
even = list(map(lambda x:x%2==0 , num))
print(even)
#map transform the every item of the list
```

```python
num  = [2,5,8,54,68,24]
even = list(filter(lambda x:x%2==0 , num))
print(even)
#filter selects items that satisfy the condition
```

```python
stu = {
    "Amit":75,
    "Neha":92,
    "Sham":88,
    "Ram":65
}
even = list(filter(lambda x:x%2==0 ,stu.values()))
print(even)
```

```python
students = [("Amit", 75), ("Neha", 92), ("Raj", 68)]
students.sort(key=lambda x:x[1])
print(students)
```

```python
students.sort(key=lambda x:x[1],reverse=True)
```

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Name": ["Amit", "Neha", "Raj"],
    "Marks": [75, 92, 35]
})
```

```python
df["results"] = df["Marks"].apply(
    lambda x : "Pass" if x>=40 else "Fail"
)
print(df)
```

```python
df.sort_values("Marks",ascending=False)
```
