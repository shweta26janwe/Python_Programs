### Question: Given sales [100, NaN, 300, 200], calculate the mean while ignoring the missing value, then replace the  missing value with that mean. 

```python
Sales = pd.DataFrame({"sales" : [100, np.nan,300,200]})
sa = sales.mean()
sa
```

```python
m_Sa = np.nanmean(sales["sales"].to_numpy())
```

```python
Sales["sales"] = Sales["sales"].filllna(m_Sa)
print("Mean sales:",m_Sa)
```

### Question: Create a DataFrame with Product, Units and Price. Calculate revenue, categorize each product as "High" when revenue is at least 500 and "Low" otherwise, and calculate total revenue. 

```python
df = pd.DataFrame(
    {
        "product":["A","B","C","D"],
        "Units" : [5,2,3,20],
        "Price" : [400,200,600,500]
    }
)
```

```python
unit = df["Units"].to_numpy()
price = df["Price"].to_numpy()
```

```python
df["Revenue"] = unit * price
df
```

```python
df["Category"] = np.where(df["Revenue"]>=500, "High","Low")
```

```python
total_rev = np.sum(df["Revenue"])
```

```python

print("Total Revenue:",total_rev)
df
```
