```python
import pandas as pd
import numpy as np
```

```python
import random 
```

```python
# 1 firset we create the empty list to store the user data
Product_Name = ["TV","Phone","Tab","PC"]
n=5
products = pd.DataFrame({
    "Product_ID" : range(1,6),
    "Product" : np.random.choice(Product_Name,n),
    # "Product" : np.random.choice(["TV","Phone","Tab","PC"],n),
    "Price" : np.random.randint(15000,100000,n),
    "Unit" : np.random.randint(15,50,n)
})
```

```python
products
```

```python
df = pd.DataFrame(products)
```

```python
df
```

```python
df["price"] = pd.to_numeric(df["Price"])
df["Unit"] = pd.to_numeric(df["Unit"])
```

```python
dff = df[(df["price"] > 50000) ]
print("Proucts whose price greater than 40000 : \n",dff)
```

```python
print("The products whose quantity is even are  \n" ,df[df["Unit"] % 2 == 0])
```
