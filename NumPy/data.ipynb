### get product input from the user and then store it in somewhere then apply filter on the price

```python
import pandas as pd
import numpy as np
```

```python
# 1 firset we create the empty list to store the user data
products= []
```

```python
# for getting user input we are using for loop
for i in range(0,5):
    name = input("Enter the name of the product ")
    unit = int(input(f"Enter the units of {name}"))
    price =int(input(f"entwr the {name}  price"))
    products.append ({
        "product" : name,
        "price":price,
        "Unit":unit})
```

```python
products
```

```python
df = pd.DataFrame(products)
```

```python
df
```

```python
df["price"] = pd.to_numeric(df["price"])
df["Unit"] = pd.to_numeric(df["Unit"])
```

```python
dff = df[(df["price"] > 50000) ]
print("Proucts whose price greater than 40000 : \n",dff)
```

```python
print("The products whose quantity is even are  \n" ,df[df["Unit"] % 2 == 0])
```
