## Missing values

```python
num = np.array([10,np.nan,20,30,40])
```

```python
np.isnan(num) 
```

```python
np.nanmean(num)  #it find mean without the nan i.e ignore nan
```

```python
np.mean(num)  #look diff between nanmean and mean 
```

```python
print(np.nanmax(num))
print(np.max(num))
print(np.nanmin(num))
print(np.min(num))
```
