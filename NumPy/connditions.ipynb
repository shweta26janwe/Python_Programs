## Concatenation, stacking and splitting

```python
a = np.array([1,2,3])
b = np.array([4,5,6])
```

```python
join = np.concatenate((a,b))
join
```

```python
print(np.vstack((a,b)))
```

```python
a = np.array([[1,2,3]])
b = np.array([[4,5,6]])
```

```python
print(np.vstack((a,b)))
```

```python
print(np.hstack((a,b)))
```

```python
print(np.concatenate((a,b)))
```

```python
a = np.array([10, 20, 30, 40, 50, 60]) 
```

```python
print(np.split(a,3))
print(np.split(a,2))
```

```python
print(np.split(a,4))
#array split does not result in an equal division
```

## keepdims, expand_dims and newaxis

```python
k = np.array([      [10,20,30],            [40,50,60]])
k
```

```python
means =k.mean(axis=1, keepdims=True)
means
```

```python
means.shape
```

```python
kmeans = k.mean(axis=1)
kmeans
```

```python
kmeans.shape
```

```python
centered = a - means
centered
```

```python
kmeans = k.mean(axis=0)
kmeans
```
