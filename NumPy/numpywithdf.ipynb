```python
import pandas as pd 
import numpy as np
```

```python
Products = {
    "product":["Laptop","phone","TV"],
    "qty" : [30,np.nan,4],
    "price" : [30000,15000,60000]
}
```

```python
Products
```

```python
df = pd.DataFrame(Products)
```

```python
df
```

```python
# Find the revenue the company

# first we need to convert the nan to 0 so we perform operation on that
units = df["qty"].to_numpy()
price = df["price"].to_numpy() #here we doesnt applt tonumy to product because we need only numercic value str not convert to num

```

```python
unit_cl = np.nan_to_num(units,nan=0.0)
price_cl = np.nan_to_num(price,nan=0.0)
```

```python
rev = unit_cl * price_cl
```

```python
rev
```

```python
df["Revenue"] =rev
```
