# Merge and Concat

```python
import pandas as pd
```

```python
emp = pd.DataFrame({
    "emp_id": [1, 2, 3, 4],
    "name": ["Amit", "Sara", "John", "Priya"],
    "dept_id": [10, 20, 10, 99],
})
dept = pd.DataFrame({
    "dept_id": [10, 20, 30],
    "dept_name": ["IT", "HR", "Finance"],
})
salary = pd.DataFrame({
    "emp_id": [1, 2, 3, 4],
    "salary": [50000, 40000, 65000, 38000],
})
```

```python
# 1. Inner join `emp` and `dept`. How many rows do you get and who is missing?
join = emp.merge(dept, on="dept_id",how="inner")
print(f"rows : {len(join)} of{len(emp)}")
join
```

```python
# 2. Left join instead, so nobody is lost.
left = emp.merge(dept, on="dept_id", how="left")
left
```

```python
# 3. Join all three tables and show name, dept_name, salary.
join3 = emp.merge(dept, on="dept_id", how="left").merge(salary, on="emp_id", how="left")
join3[["name", "dept_name", "salary"]]
```

```python
# 4. Find employees whose `dept_id` does not exist in `dept` (use `indicator`).
check = emp.merge(dept, on="dept_id",how="outer" ,indicator=True )
print(check)
print()
print("Emp with wrong dept_id:")
print(check[check["_merge"] == "left_only"][["name","dept_id"]])
```

```python
# 5. Find departments that have no employee.
# dcheck = emp.merge(dept, on="dept_id", how="outer", indicator=True)
print("Dept which have no emp")
check[check["_merge"] == "right_only"]
```

```python
# 6. Stack `emp` with a new row for employee 5 (`Ravi`, dept 20) using `concat`.
# new_emp = pd.DataFrame({"emp_id":[5],"name":["Ravi"], "dept_id" :[20]})
new_emp = pd.DataFrame({"emp_id": [5], "name": ["Ravi"], "dept_id": [20]})
emp_all = pd.concat([emp, new_emp], ignore_index=True)
emp_all
```

```python
# 7. Total salary per department name.
(emp.merge(dept,on="dept_id",how="inner").merge(salary,on="emp_id",how="inner").groupby("dept_name")["salary"].sum())
```
