```python
# map() : Combines elements from two or more iterables.
num = [1,2,3,4]
res = map(lambda x: x*2,num)
print(list(res))
```

```python
# Q1. Multiply every number by 5:
ress = map(lambda x : x*5, num)
print(list(ress))
```

```python
# Find the square of every number:
sq =[]
for i in num:
    sq.append(pow(i,2))
print(sq)

ress = map(lambda x: pow(x,2) ,num)
print(list(ress))
```

```python
# filter() selects elements that satisfy a condition.
ress = map(lambda x: pow(x,2) == 16 ,num)
print(list(ress))
```

```python
ress = filter(lambda x: pow(x,2) == 16 ,num)
print(list(ress))
```

```python
names = ["Raj", "Shweta", "Amit", "Prakash", "Neha"]
filter_name = filter(lambda x : len(x)>5,names)
print(list(filter_name))
```

```python
# all() : all() returns True if all elements are truthy; otherwise False.
print(all(i%2 == 0 for i in num))
```

```python
# any() returns True if at least one element is truthy; otherwise False.
print(any(i%2 == 0 for i in num))
```
