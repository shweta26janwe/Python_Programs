# Columns Method

```python
import pandas as pd
import numpy as np
```

```python
emp = pd.DataFrame({
    "name": ["amit sharma", "SARA KHAN", "john dsouza", "priya rao", "ravi kumar"],
    "dept": ["IT", "HR", "IT", "SALES", "hr"],
    "salary": [50000, 40000, 95000, 38000, 45000],
    "joining_date": ["2020-01-15", "2019-06-01", "2015-03-20", "2023-11-05", "2021-08-30"],
})
```

```python
# 1. Clean `name` (title case) and `dept` (upper case).
emp["name"] = emp["name"].str.strip().str.title()
emp["dept"] = emp["dept"].str.upper()
emp
```

```python
# 2. Split `name` into `first_name` and `last_name`.
emp[["first_Name","Last_Name"]] = emp["name"].str.split(" ",expand=True)
emp.head()
```

```python
# 3. Add `annual_salary` (salary * 12) and `tax` (10% of annual salary).
emp["annual_salary"] = emp["salary"] * 12
emp["tax"] = 10 * emp["annual_salary"]/100
```

```python
emp
```

```python
# 4. Add a column `level`: above 90000 = "Senior", above 45000 = "Mid", else "Junior".


emp["level"] = np.select([emp["salary"] > 90000,emp["salary"] > 45000],["Senior","Mid"],default="Junior")
```

```python
emp
```

```python
# 5. Map the department to a head: IT -> "Mr. Rao", HR -> "Ms. Iyer", SALES -> "Mr. Shah".
dept_head = {"IT":"Mr. Rao", "HR":"Ms. Iyer", "SALES":"Mr. Shah"}
emp["dept_head"] = emp["dept"].map(dept_head)
```

```python
emp
```

```python
# 6. Convert `joining_date` to a date and add `joining_year` and `joining_month_name`.
emp["joining_date"] = pd.to_datetime(emp["joining_date"])
emp["joining_year"] = emp["joining_date"].dt.year
emp["joining_month_name"] = emp["joining_date"].dt.month_name()
emp[["name", "joining_date", "joining_year", "joining_month_name"]]
```

```python
# 7. Add `years_worked` = whole years between joining date and 2026-09-23.
emp["years_worked"] = ((pd.Timestamp("2026-09-23") - emp["joining_date"]).dt.days / 365).astype(int)
```

```python
emp[["name", "joining_date", "joining_year", "joining_month_name", "years_worked"]]
```

```python
# 8. Add a salary band using `pd.cut`: 0-40000 "Low", 40000-70000 "Medium", 70000+ "High".
pd.cut(emp["salary"], bins=[0,40000,70000,1000000],labels=["Low","Mwdium","High"])
```
