How to count the word in each sentence

```python
sentence = "India is a great great country"
```

```python
words = sentence.split()
print(words)
```

```python
for word in words:
    print(f"{word} : {len(word)}")

# In above for loop we find the length each word in this sentence
```

```python
#basic way to count the word
count = {}
for word in words:
    if word in count:
        count[word] += 1
    else:
        count[word] = 1
print(count)
```

```python
print("The count of each word in sentence",{sentence})
for word in set(words):
    print(f"{word} : {words.count(word)}")

```

```python
dict = {}
for word in words:
    dict[word] = dict.get(word,0)+1

print(dict)
```
