### create a new list containing only numbers greater than 15.

```python
numbers = [10, 25, 14, 30, 8, 19, 15]
newli = list()
for i in numbers:
    if i > 15:
        newli.append(i)
print("New list which contain only no. > 15 list are:",newli)
```

### Count the numbers that are divisible by 3 but not divisible by 2.

```python
numbers = [3, 6, 9, 12, 15, 18, 21, 24, 27]
count =0 
for i in numbers:
    if i%3==0 and i%2!=0:
        count += 1

print("Count is ",count)
```

### Find the largest even number in the list without using max().

```python
numbers = [12, 7, 24, 15, 32, 9, 40, 18, 5]
maxi = float('-inf')
for i in numbers:
    if i%2==0:
        if i>maxi:
            maxi=i
print("Even MAximum value is ",maxi)

#for minimum allocate float('inf')
```

### : Find the second-largest number in the list without using

```python
numbers = [12, 7, 24, 15, 32, 9, 40, 18, 5]
maximum = float('-inf')
secmax = float('-inf')
for i in numbers:
    if i>secmax:
        if i > maximum:
            secmax = maximum
            maximum = i
        else:
            secmax = i
        
print("second max is ",secmax)
        

```

###  Find the smallest odd number

```python
numbers = [12, 7, 24, 15, 32, 9, 40, 18, 5]
mini = float('inf')
for i in numbers:
    if i%2 != 0 and i<mini:
        mini =i 

print("Minimum no. is ",mini)
```
