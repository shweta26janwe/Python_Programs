## Dictionary comprehension 

###  Create a dictionary where each number is the key and its square is the value. 

```python
numbers = [1, 2, 3, 4] 
squares = { n:n**2 for n in numbers}
print(squares)
```

```python
words = ["data", "python", "AI"]  
lengths = { word:len(word) for word in words}
print("The length of each word is ",lengths)
```

```python
words = ["data", "python", "AI"]  
lengths = { word:type(word) for word in words}
print("The length of each word is ",lengths)
```

###  Create a dictionary of students whose score is at least 60. 

```python
scores = {"Asha": 82, "Ravi": 48, "Lee": 67} 
passed = {name:score for name,score in scores.items() if score>= 60}
print(passed)
```

```python
scores = {"Asha": 82, "Ravi": 48, "Lee": 67} 
passed = {(score) for name,score in scores.items() if score>= 60}
print(passed)
```

```python
print(max(scores))
print(max(scores.keys()))
print(max(scores.values()))
```

```python
scores = {"Asha": 82, "Ravi": 48, "Lee": 67} 
passed = {(name,score) for name,score in scores.items() if score>= 60}
print(passed)
```

### Create a dictionary of product prices after a 10% discount. 

```python
prices = {"pen": 100, "book": 500, "bag": 1000} 
disc={product:(price-(price*0.10)) for product,price in prices.items()}
print(disc)
```

###  Swap values and keys when original values are unique and hashable. 

```python
codes = {"Pune": "PNQ", "Mumbai": "HYD", "Delhi": "DEL","Mumbai":"BOM","Mu":"DEL"} 
reverse = {val:city for city,val in codes.items()}
print(reverse)
# see in this code if key or value are repeated the it overwrites
```

###  Map each number to the label 'even' or 'odd'. 

```python
numbers = [1, 2, 3, 4] 
labels = { num : "Even" if num%2==0 else "Odd" 
          for num in numbers}
print(labels)
```

###  Build a dictionary of sensor IDs and readings, keeping only positive readings. 

```python
readings = {"S1": 0.2, "S2": -1.1, "S3": 3.4} 
posreadings = {sensor:val if val>0 else 0 
               for sensor,val in readings.items() }
print(posreadings)
```

### Extract the domain portion after @ from email addresses. 

```python
emails = ["ana@example.com", "raj@company.in"] 
domains = {email:email.split("@")[-1] for email in emails}
print(domains)
```

```python
emails = ["ana@example.com", "raj@company.in"] 
domains = {email : (email.split("@")[-1])[-3:] for email in emails}
print(domains)
```

```python
emails = ["ana@example.com", "raj@company.in"] 
domains = {email : (email.split("@")[-1])[0:-4] for email in emails}
print(domains)
```

### Use corresponding items from two lists as keys and values. 

```python
names = ["Asha", "Ravi", "Lee"] 
scores = [82, 74, 91,78] 
res = {}
for i in range(len(names)):
    res[names[i]] = scores[i]
print(res)
```

```python
print(zip(names,scores))
```

```python
#using dict comp
result ={name:score for name,score in zip(names,scores)}
print(result)
```

```python
names = ["Asha", "Ravi", "Lee"] 
scores = [82, 74, 91,78]
result ={name:score for name,score in zip(names,scores)}
print(result)
```

### Round every product metric to two decimal places.

###  Understand why a dictionary comprehension is not the best tool for counting repeated items. 

```python
items = ["red", "blue", "red"] 
counts = {}
for item in items:
    counts[item] = counts.get(item,0)+1
counts

```

```python
count = {item:items.count(item) for item in items}
count
```

###  Create a dictionary that flags metrics as 'strong' when they are at least 0.90, otherwise 'review'. 

```python
metrics = {"precision": 0.94, "recall": 0.87, "f1": 0.91} 
flags = {metric:"Strong" if val >= 0.90 else "review"
         for metric,val in metrics.items() }
flags
```
