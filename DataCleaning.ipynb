# Data Cleaning

```python
import pandas as pd
```

```python
students = pd.DataFrame({
    " Roll No ": [1, 2, 3, 4, 4, 5],
    "Student Name": [" amit", "SARA ", "john", "priya", "priya", None],
    "City": ["pune", "PUNE", "delhi", None, None, "Mumbai"],
    "Marks": ["85", "92", "abc", "65", "65", "70"],
    "Joined": ["2026-01-05", "2026-01-06", "07-01-2026", "2026-01-08", "2026-01-08", None],
})
```

```python
# 1. Print how many missing values each column has.
print("Number of null value in each col")
students.isna().sum()
```

```python
# 2. Remove the duplicate row.
print("Before drop duplicate :", len(students))
stu = students.drop_duplicates(subset=[" Roll No "],keep=False) #both remove
print("After drop duplicate :", len(stu))
```

```python
# 3. Clean the column names (no spaces, small letters, underscore).
students.columns = students.columns.str.strip().str.lower().str.replace(" ","_")
print(list(students.columns))
print(students.head())
```

```python
# 4. Clean the student names and cities (no extra spaces, title case).
students["student_name"] = students["student_name"].str.strip().str.title()
print(students[["student_name","city"]])
```

```python
# 5. Convert `marks` to a number (bad values become missing), then fill the missing marks
#    with the average.
students["marks"] = pd.to_numeric(students["marks"],errors="coerce")
students["marks"] = students["marks"].fillna(students["marks"].mean())
students["marks"]
```

```python
# 6. Convert `joined` to a date.
students = students.rename(columns = {"joined":"date"})
students.columns
students["date"] = pd.to_datetime(students["date"],errors="coerce",format="mixed",dayfirst=True)
students.head()
```

```python
# 7. Fill the missing city with `"Unknown"`.
students["city"] = students["city"].fillna("unknown")
students["city"]
```

```python
# 8. Print the final table and `info()`.
print(students.head())
students.info()
```
