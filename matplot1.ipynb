# Lineplot of Monthly Sales 

```python
import datetime as dt
```

```python
df["Order_Date"].dtypes
```

```python
#to extract the month and year we have to first convert it into datetime type
df["Order_Date"] = pd.to_datetime(df["Order_Date"], format="%d-%m-%Y",errors="coerce")
```

```python
print(df["Order_Date"].dtype)
```

```python
df
```

```python
dff = df["Order_Date"].dt.year
```

```python
dff
```

```python
df["Month"] = df["Order_Date"].dt.month
df["Year"] = df["Order_Date"].dt.year
```

```python
df
```

```python
monthly_Sales = (
    df.groupby(["Month","Year"])["Sales_Amount"].sum().sort_index()
)
```

```python
monthly_Sales
```

```python
monthly_Sales = (
    df.groupby(["Month","Year"])["Sales_Amount"].sum().sort_index(ascending=False)
)
```

```python
monthly_Sales
```

```python

plt.figure(figsize=(15,8))

plt.plot(
    monthly_Sales.index.map(lambda x: f"{x[0]}-{x[1]}"),
    monthly_Sales.values
)
plt.title("monthly_Sales")
plt.xlabel("date")
plt.ylabel("sales amount")

plt.tight_layout()
plt.show()


```

```python
plt.figure(figsize=(15,8))

plt.plot(
    monthly_Sales.index.map(lambda x: f"{x[0]}-{x[1]}"),
    monthly_Sales.values,
    marker="o"
)
plt.title("monthly_Sales")
plt.xlabel("date")
plt.ylabel("sales amount")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
