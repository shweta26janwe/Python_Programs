```python
# Import ABC and abstractmethod.
# Create:
# class Animal(ABC):
# Create an abstract method:
# sound()
# Create a child class Dog.
# Dog should implement sound().
# It should print: bark

from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        print("dog barking")

dog = Dog()
dog.sound()
```

```python
# Create:
# Shape
#   ↓
# Rectangle
#   ↓
# area()

# Requirements:
# Parent:
# class Shape(ABC):
# Create abstract method:
# area()
# Child: rectangle
import math
class Shape(ABC):
    def __init__(self,*args):
        self.args = args
        

    @abstractmethod
    def area(self):
        pass

class Rectangle(Shape):
    def area(self):
        print(f"Area of rectangle : {self.args[0]*self.args[1]}sq")

class Square(Shape):
    def area(self):
        print("Area of square is ",self.args[0]*self.args[0]) 

class Circle(Shape):
    def area(self):
        print("Area of circle is ", math.pi * self.args[0] * self.args[0])

a = Rectangle(3,7)
a.area()
s = Square(5)
s.area()
c = Circle(3)
c.area()
```

```python
# Create:
# Payment
#    ↓
# CardPayment
#    ↓
# pay()

# Parent:
# class Payment(ABC):

# Abstract method:
# pay(amount)

# Child:
# class CardPayment(Payment):

class Payment(ABC):
    @abstractmethod
    def pay(self,amt):
        pass
    #concret mrthod
    def receipt(self,amt):
        print("Receipt generated for ", amt)
class CardPayment(Payment):
    def pay(self,amt):
        print("Payment by Card is ", amt)

p = CardPayment()
p.pay(500)
p.receipt(500)
```

```python
#Payment

class DataPipeline(ABC):

    @abstractmethod
    def extract(self):
        pass

    @abstractmethod
    def transform(self,data):
        pass

    @abstractmethod
    def load(self,data):
        pass

    def run(self):
        data = self.extract()
        print(data)
        data = self.transform(data)
        print(data)
        data = self.load(data)

class CSVPipeline(DataPipeline):
    def extract(self):
        return "Extracting file"
    def transform(self,data):
        return "Transforming file"
    def load(self,data):
        print("Loading file " , data)

e = CSVPipeline()
e.run()
```
