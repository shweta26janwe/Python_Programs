### Create a Student Class

Create a class named Student with the following requirements:
1. Create an __init__() method that accepts name, age, and marks.
2. Store these values as instance variables.
3. Create a method called display_details() that prints all student details.
4. Create two objects with different student details.
5. Call the display_details() method for both objects.

```python
class student:
    def __init__(self,name,age,marks):
        self.name = name
        self.age = age
        self.marks = marks

    def display_details(self):
        print(f"The info of student {self.name} is \n Age: {self.age} \n Marks : {self.marks}")


s1 = student("Shweta",34,89)
s2 = student("Shruti",24,79)


s1.display_details()
s2.display_details()
```

### Create a class named Employee.
1. Create a constructor __init__() that accepts:
   - name
   - salary (monthly salary)
   - department
2. Store these values using instance variables.
3. Create a method called annual_salary() that calculates and displays the annual salary.
4. Create another method called display_details() that displays:
   - Employee name
   - Department
   - Annual salary
5. Create two employee objects and call their methods.

```python
class Employee:
    def __init__(self,name,salary,dept):
        self.name = name
        self.sal = salary
        self.dept = dept

    def annual_sal(self):
        annuall_sal = 12 * self.sal
        return annuall_sal

    def display_details(self):
        print(f"The Details of Emp is \nName : {self.name}\nDepartment : {self.dept}\nAnnual Salary : {self.annual_sal()} \n")


e1 = Employee("Manoj Pade",45000,"Sales")
e2 = Employee("Pravin Ghate",35000,"HR")

e1.display_details()
e2.display_details()
```

### Create a class named BankAccount.
Create a constructor __init__() that accepts:
- account_holder
- balance



Create a method deposit(amount):
- Add the deposited amount to the current balance.



Create a method withdraw(amount):
- Subtract the amount if sufficient balance is available.
- Otherwise, display "Insufficient balance".



Create a method check_balance():
- Display the current account balance.

```python
class Bank_Acc:
    def __init__(self,acc_holder,balance):
        self.acc_holder = acc_holder
        self.bal = balance

    def deposit(self,amt):
        self.bal =  self.bal + amt
        print("The Deposited amt ",amt)
        print("The Available amt ",self.bal)

    def withdraw(self,amt):
        if self.bal >= amt :
            self.bal =  self.bal - amt
            print("The withdraw amt ",amt)
            print("The Available amt ",self.bal)
        else:
            print("Insufficient Balance")

    def check_bal(self):
        print("The current balance in account is ",self.bal)

a1 = Bank_Acc("Gajanan",40000)
a2 = Bank_Acc("Mohini Junghare",30000)

a1.withdraw(3000)
a1.check_bal()
a1.deposit(250)
```

### Create a Vehicle and Car class
Requirements:
1. Create a parent class Vehicle.
2. Add a constructor that accepts:
   - brand
   - model
   - year
3. Create a method display_vehicle() to display these details.
4. Create a child class Car that inherits from Vehicle.
5. Add an additional attribute fuel_type in the Car class.
6. Create a method display_car() to display all vehicle details, including fuel type.

```python
class Vehicle:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

    def display_vehicle(self):
        print("Vehicle Details:")
        print(f"Brand: {self.brand}")
        print(f"Model: {self.model}")
        print(f"Year: {self.year}")


class Car(Vehicle):
    def __init__(self, brand, model, year, fuel):
        super().__init__(brand, model, year)
        self.fuel_type = fuel

    def display_car(self):
        self.display_vehicle()
        print(f"Fuel Type: {self.fuel_type}")

c1 = Car("Honda", "City", 2026, "Petrol")

c1.display_car()
```
