###  Normalize each city string by removing surrounding spaces and converting to lowercase

```python
cities = [" Pune ", "MUMBAI", " Delhi "] 
Norm_cities = [ c.strip() for c in cities]
```

###  Convert a list of lists into one list. 

```python
groups = [[1, 2], [3, 4], [5]] 
n_grp = []
for i in groups:
    for j in i:
        n_grp.append(j)
print(n_grp)
```

```python
#using list comprehension
new_li = [j for i in groups for j in i]
new_li
```

###  Replace missing-like sentinel values 
### Create a new list where -999 is replaced by None. 

```python
readings = [23, -999, 25, -999, 21] 
new_Readings = [None if r==-999 else r for r in readings ]
new_Readings
```
