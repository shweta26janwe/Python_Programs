```python
#second max element 
numbers = [12, 45, 23, 45, 67, 23, 54]
print(max(numbers))
```

```python
max = float('-inf')
secmax = float('-inf')
for i in numbers:
    if i > secmax:
        if i > max:
            secmax = max
            max = i
        else:
            secmax = i
print(secmax)
```

```python
# Given a list, remove duplicate elements without changing the original order. Do not use set().

numbers = [10, 20, 10, 30, 20, 40, 30, 50]
noduplicates =[]
for i in numbers:
    if noduplicates.count(i) < 1 :
        noduplicates.append(i)
print(noduplicates)
```

```python
numbers = [10, 20, 10, 30, 20, 40, 30, 50]
noduplicates =[]
for i in numbers:
    if i not in noduplicates:
        noduplicates.append(i)
print(noduplicates)
```

```python
#Count the frequency of each fruit without using Counter.
fruits = [
    "apple", "banana", "apple", "orange",
    "banana", "apple", "mango", "orange"
]
fruitsfreq = {}
for fruit in fruits:
    fruitsfreq[fruit] = fruitsfreq.get(fruit,0)+1

print(fruitsfreq)
```

```python
# Separate even and odd numbers
numbers = [12, 7, 9, 24, 36, 15, 18, 21, 40]
even = list()
odd = list()
for i in numbers:
    if i%2 == 0:
        even.append(i)
    else:
        odd.append(i)

print(even)
print(odd)
```

```python
#Find common elements between two lists
list1 = [10, 20, 30, 40, 20, 50]
list2 = [20, 40, 60, 20, 70, 30]
common = []
for i in list1:
    if i in list2 and i not in common:
        common.append(i)


print(common)
```

```python
#Count word frequency
import string
sentence = "Python is easy. Python. is powerful, and python is popular."
wordcnt = {}
sentence = sentence.translate(str.maketrans("","", string.punctuation))
for sen in sentence.lower().split():
    wordcnt[sen] = wordcnt.get(sen,0)+1
print(wordcnt)
```

```python
#Check whether two strings are anagrams
str1 = "Listen"
str2 = "Silent"
if sorted(str1.lower()) == sorted(str2.lower()):
    print(True)
else:
    print(False)
```

```python
#. Sort a dictionary by values in descending order
students = {
    "Amit": 75,
    "Priya": 92,
    "Rahul": 68,
    "Sneha": 85,
    "Neha": 92
}
sortedtuple = dict(sorted(students.items(),key=lambda x : x[1],reverse=True))
print(sortedtuple)
```

```python
#Given these two dictionaries, combine the sales for each city. If a city appears in both dictionaries, add the values.
sales_january = {
    "Pune": 12000,
    "Mumbai": 18000,
    "Nagpur": 9000
}

sales_february = {
    "Pune": 15000,
    "Mumbai": 10000,
    "Nashik": 7000
}
combine = sales_january.copy()
for city,feb_sales in sales_february.items():
    if city in sales_january:
        combine[city] = combine.get(city,0)+ feb_sales
    

print(combine)
```
