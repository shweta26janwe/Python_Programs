## Set Comprehension

```python
#Create a set of squares from numbers that may contain duplicates. 
numbers = [1, 2, 2, 3, 3, 3] 
sqaures = { x**2 for x in numbers}
```

```python
sqaures
```

```python
# Normalize names to lowercase and keep unique results. 
names = ["Asha", "ASHA", "Ravi", "ravi"]
uniNames = { n.lower() for n in names}
uniNames 
```

```python
# Find the distinct lengths of words. 
words = ["AI", "data", "model", "ML"] 
len = { len(w) for w in words}
len
```

```python
# Create a set of unique even numbers. 
numbers = [2, 2, 3, 4, 4, 5, 6] 
even = { i for i in numbers if i%2 == 0}
```

```python
even
```

```python
#Get distinct first letters from non-empty words. 
words = ["data", "deep", "AI", "", "analytics"] 
distinctFirst = {w[0].lower() for w in words if w}
distinctFirst
```

```python
# Extract unique extensions from filenames. 
files = ["report.csv", "sales.csv", "model.pkl", "notes.docx"] 
ext = {f.split(".")[1] for f in files if f}
ext
```

```python
# Keep distinct positive readings. 
values = [-1, 0, 2, 2, 3, -1, 4] 
distPos = {v for v in values if v >0}
distPos
```

```python
# Round measurements to one decimal place and keep distinct rounded values.
measurements = [1.24, 1.26, 2.01, 2.04] 
rounded = {round(m,1) for m in measurements}
```

```python
rounded
```

```python
# Collect unique lowercase words from sentences. 
sentences = ["AI helps data", "Data builds AI"] 

words = [sentence.split() for sentence in sentences 
         ]
words
```

```python
word = { word.lower() for sentence in sentences for word in sentence.split()}
word
```

```python
#Get distinct two-character prefixes from words of length at least two.
words = ["data", "database", "AI", "a"] 
prefixes = {w[:2].lower() for w in words if len(w)>=2}
prefixes
```

```python
#Create a set of unique codes excluding zero. 
codes = [101, 0, 102, 101, 103, 0] 
uniqueCodes = { c for c in codes if c!=0}
uniqueCodes
```

```python
# Create unique pairs of category and region. 
records = [("food", "west"), ("food", "west"), ("travel", "east")] 
pairs = {pair for pair in records}
pairs
```

```python
pairss = {(category,region) for category,region in records}
print(pairss)
regions = {(region) for category,region in records}
print("Only regions :",(regions))
```

```python
# Keep unique numbers divisible by three. 
numbers = [3, 6, 6, 7, 9, 12, 12] 
num = {n for n in numbers if n%3==0}
num
```

```python
#Strip spaces, lowercase labels, discard blanks, and keep unique labels. 
labels = [" Retail ", "retail", "AI ", "", " ai"] 
cleanlabel = { label.strip().lower() for label in labels if label} 
cleanlabel
```

```python
labels = [" Retail ", "retail", "AI ", "", " ai"] 
cleanSet = {label.strip().lower() for label in labels if label!=""}
print("Cleaned data : ",cleanSet)
#we can also written in this way also to tackle with blank space
cleanSet = {label.strip().lower() for label in labels if label.strip()}
print("Cleaned data : ",cleanSet)
```
